```python
import math 
import random
```

1- TEK NÖRON FOWRARD PASS

```python
#rastgele 3 adet girdi belirliyorum.
a, b, c = [random.random() for i in range(3)]
print("a = ", a)
print("b = ", b)
print("c = ", c)

#daha sonra bu girdiler için rastgele weight değerleri belirliyorum.
w1_a, w1_b, w1_c = [random.uniform(-1, 1) for i in range(3)]
print("w1_a = ", w1_a)
print("w1_b = ", w1_b)
print("w1_c = ", w1_c)


#rastgele bir bias değeri seçiyorum.
bias1 = 2

#wehighted sum hesaplaması yapıyorum.
weighted_sum = (a * w1_a) + (b * w1_b) + (c * w1_c) + bias1

print("weighted_sum = ", weighted_sum)

#daha sonra bu weighted sum değerini sigmoid fonksiyonuna sokuyorum. Hazır fonksiyon kullanmak yerine kendim yazıyorum.
sigmoid_weighted_sum = 1/(1+math.exp(-weighted_sum))

print("sigmoid_weighted_sum = ", sigmoid_weighted_sum)

```

2- ÇOK NÖRONLU KATMAN

```python
#2 nöron daha oluşturmak için 2 adet daha bias değeri belirliyorum. Aynı zamanda her girdi için 3 weight değeri olacağı için 6 adet weight değeri daha belirliyorum.

bias2 = 3
bias3 = 4

w2_a, w2_b, w2_c = [random.uniform(-1, 1) for i in range(3)]
w3_a, w3_b, w3_c = [random.uniform(-1, 1) for i in range(3)]

#daha sonra belirlediğim bu bias ve weight değerlerini kullanarak 2 nöron için weighted sum hesaplaması yapıyorum.

weighted_sum2 = (a * w2_a) + (b * w2_b) + (c * w2_c) + bias2
weighted_sum3 = (a * w3_a) + (b * w3_b) + (c * w3_c) + bias3
#Bu weighted sum değerlerini de sigmoid fonksiyonuna sokuyorum.
sigmoid_weighted_sum2 = 1/(1+math.exp(-weighted_sum2))
sigmoid_weighted_sum3 = 1/(1+math.exp(-weighted_sum3))


print("weighted_sum2 = ", weighted_sum2)
print("sigmoid_weighted_sum2 = ", sigmoid_weighted_sum2)
print("weighted_sum3 = ", weighted_sum3)
print("sigmoid_weighted_sum3 = ", sigmoid_weighted_sum3)
```

3- BASİT LOSS FONKSİYONU YAZMA

```python
#Rastgele bir target değeri belirliyorum. Bu değer sigmoid fonksiyonundan dolayı 0 ile 1 arasında olacak. Bu değer ile sigmoid fonksiyonundan çıkan değerlerin farkını loss fonksiyonu olarak kullanacağım. Loss fonksiyonu olarak Mean Squared Error (MSE) kullanacağım.
#Öğrenme amaçlı bir çalışma olduğu için tek nöron için loss fonksiyonu hesaplaması yapacağım.

Target = 0.5
tahmin = sigmoid_weighted_sum
print("tahmin = ", tahmin)
loss = (Target - tahmin) ** 2

print("Target = ", Target)
print("loss = ", loss)

```

4- LOSS EĞRİSİ

```python
#Parametreler ile değişim yapabilmek için bu kısımda yeni bir nöron oluşturuyorum ve değerlerini manuel olarak random belirliyorum. Daha sonra bu nöron için loss fonksiyonu hesaplaması yapacağım. 
#Daha sonra parametreleri değiştirip loss fonskiyonunun değişimini bir listeye kaydedeceğim daha sonra bu listeyi kullanarak loss fonksiyonunun değişimini görselleştireceğim.
```

```python
import matplotlib.pyplot as plt

girdi1= 0.31
girdi2= 0.57
girdi3= 0.42

girdi1_weight = -0.23
girdi2_weight = 0.12
girdi3_weight = -0.45

bias_test = 1.5

target_test = 0.5

w_walues = []
loss_values = []

for i in range(100):
    weighted_sum_test = (girdi1 * girdi1_weight) + (girdi2 * girdi2_weight) + (girdi3 * girdi3_weight) + bias_test
    sigmoid_weighted_sum_test = 1/(1+math.exp(-weighted_sum_test))
    tahmin_test = sigmoid_weighted_sum_test
    loss_test = (target_test - tahmin_test) ** 2

    print("Iteration: ", i)
    print("weighted_sum_test = ", weighted_sum_test)
    print("sigmoid_weighted_sum_test = ", sigmoid_weighted_sum_test)
    print("loss_test = ", loss_test)

    w_walues.append((girdi1_weight,))
    loss_values.append(loss_test)
    

    #Ağırlık Parametrelerini değiştiriyorum.
    girdi1_weight -= 0.1 #ilk başta "+" olarak belirledim istediğim parabolü yakalayamadığım için "-" olarak değiştirdim.
    #girdi2_weight += random.uniform(-0.1, 0.1)
    #girdi3_weight += random.uniform(-0.1, 0.1)
    

    

plt.figure(figsize=(12, 4))
plt.plot(w_walues, loss_values, marker='o')
plt.xlabel('Weights')
plt.ylabel('Loss')
plt.title('Weight Updates vs Loss')
plt.grid()
plt.show()
```

```python
#Bir parametre değerini değiştirip değişim oranına göre oranını alıp o noktanın sayısal türevini almış olacağız böylelikle az 
# önce yaptığımız çalışmanın sonucundaki ortaya çıkan parabolün o noktadaki eğimini bulup o paramtereyi hangi yönde değiştireceğimizi bulacağız.
girdi1= 0.31
girdi2= 0.57
girdi3= 0.42
girdi1_weight = -0.23
girdi2_weight = 0.12
girdi3_weight = -0.45
bias_test = 1.5
target_test = 0.5
weighted_sum_test = (girdi1 * girdi1_weight) + (girdi2 * girdi2_weight) + (girdi3 * girdi3_weight) + bias_test
sigmoid_weighted_sum_test = 1/(1+math.exp(-weighted_sum_test))
tahmin_test = sigmoid_weighted_sum_test
loss_test = (target_test - tahmin_test) ** 2
h = 0.01  # Küçük bir değişim oranı
girdi1_weight_plus_h = girdi1_weight + h
weighted_sum_test_plus_h = (girdi1 * girdi1_weight_plus_h) + (girdi2 * girdi2_weight) + (girdi3 * girdi3_weight) + bias_test
sigmoid_weighted_sum_test_plus_h = 1/(1+math.exp(-weighted_sum_test_plus_h))
tahmin_test_plus_h = sigmoid_weighted_sum_test_plus_h
loss_test_plus_h = (target_test - tahmin_test_plus_h) ** 2
turev = (loss_test_plus_h - loss_test) / h
print(f"Numerical derivative (turev): {turev:.3f}")
```
