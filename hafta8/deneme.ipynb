```python
import requests

url = "https://dn721308.ca.archive.org/0/items/ahmet-hamdi-tanpinar-saatleri-ayarlama-enstitusu/Ahmet%20Hamdi%20Tanpınar%20-%20Saatleri%20Ayarlama%20Enstitüsü_djvu.txt"
# Bu çalışmazsa attığın orijinal ia600401... linkini yaz.

r = requests.get(url, timeout=60)
r.raise_for_status()
metin = r.content.decode("utf-8", errors="replace")

with open("kurk_mantolu_madonna.txt", "w", encoding="utf-8") as f:
    f.write(metin)

print("Toplam karakter:", len(metin), "| Farklı karakter:", len(set(metin)))
print(metin[:500])
```
